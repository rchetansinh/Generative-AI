```python
# Create a list to hold all the cleaned messages
corpus = []

for i in range(len(messages)):
    # 1. Remove special characters (Fixed regex to a-zA-Z)
    review = re.sub("[^a-zA-Z]", ' ', messages['message'][i])
    
    # 2. Convert to lowercase
    review = review.lower()
    
    # 3. Tokenize the entire sentence at once to get accurate POS tags
    tokens = word_tokenize(review)
    # 4. Get POS tags for the whole sentence
    tags = pos_tag(tokens)
    
    # 5. Lemmatize words using their tags
    cleaned_words = []
    for word, tag in tags:
        # Determine noun, verb, adjective, or adverb
        if tag.startswith('V'):
            pos = 'v'
        elif tag.startswith('J'):
            pos = 'a'
        elif tag.startswith('R'):
            pos = 'r'
        else:
            pos = 'n'  # Default to noun
            
        cleaned_words.append(wlemmatizer.lemmatize(word, pos=pos))
    
    # 6. Rejoin the words back into a clean string and add to our corpus
    corpus.append(" ".join(cleaned_words))

# Save the results back into your dataframe
#messages['cleaned_message'] = corpus

print(corpus)
```

```python
tfidf=TfidfVectorizer(max_features=100,ngram_range=(1,2))
X= tfidf.fit_transform(corpus).toarray()
```
