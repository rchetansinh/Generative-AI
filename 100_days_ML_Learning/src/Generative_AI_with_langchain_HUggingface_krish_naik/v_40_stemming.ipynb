```python
##classifincation problem
## comments in product review is positive review or negative review
## Reviews will contain this words  like : eat , eaten , eating 
## go , gone going
```

```python
from nltk.stem import PorterStemmer
words = ["eat","eaten","eating","writing","writes","Programming","program","history","congratulations"]
```

```python
steming = PorterStemmer()
for word in words:
    print(f"{word}------>stemword---->{steming.stem(word)}")

#here only one issue is there are some word it stemed wrongly like history that we can fix
#lemmatization
```

```python
#Regesxpstemmer
from nltk.stem import RegexpStemmer
```

```python
reg_stemmer= RegexpStemmer('ing$ | s$ | e$ | able$',min=4)
reg_stemmer
```

```python
reg_stemmer.stem("eating")
```

```python
reg_stemmer.stem("ingeating")
```

```python
### Snowball stemmer
from nltk.stem import SnowballStemmer
```

```python
snowballstemmer= SnowballStemmer(language="english")
```

```python
for word in words:
    print(f"{word} ----> {snowballstemmer.stem(word)}")
```
