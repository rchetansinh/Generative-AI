```python
lamatizer= WordNetLemmatizer()
words = ["eat","eaten","eating","writing","writes","Programming","program","history","congratulations"]
```

```python
#POS-Noun - n 
# verbs -v
#adjective - a
# adverb-r
lamatizer.lemmatize("going",pos="v")
```

```python
for word in words:
    print(word + "--->" + lamatizer.lemmatize(word,pos="n"))
```

```python
for word in words:
    print(word + "--->" + lamatizer.lemmatize(word,pos="v"))
```

```python
from nltk.stem import WordNetLemmatizer
from nltk import pos_tag, word_tokenize
import nltk

nltk.download('averaged_perceptron_tagger_eng')

lemmatizer = WordNetLemmatizer()
text = "The dogs were running"
tokens = word_tokenize(text)
print(tokens)
tags = pos_tag(tokens)
print(tags)

for word, tag in tags:
    # Map NLTK tags to WordNet tags (n, v, a, r)
    pos = 'n' if tag.startswith('N') else 'v' if tag.startswith('V') else 'n'
    print(f"{word} -> {lemmatizer.lemmatize(word, pos=pos)}")   
```
